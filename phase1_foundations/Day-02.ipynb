```python
#FulLArithematic set


principal = 100
rate = 0.07
years = 10

# * is for multiplication and ** is for exponentiation
# 10**2 will give 100
future_value = principal * (1 + rate) ** years
print(round(future_value, 2))   # 196.72
```

```python
# Associativity and precedence

#Python reads as -(2**2) because ** has precedence over -(unary)
print(-2 ** 2)     # -4, not 4!
print((-2) ** 2)    # 4

# ** is right associative, python reads is right to left
print(2 ** 3 ** 2)   # 512, not 64
```

```python
# Comparison Operators

print(2 > 3)

print(2 <= 3)

print(2 >= 2)


# even if one is True it is True overall
print(3 < 4 or 4 < 2)


# This means both sides should agree for it to be True
print(3 < 4 < 2)


# <= reads as less than or equal to if reading from left,
#          equal to or greater than if reading from the right 
print(4 <= 5 >= 2)
```

```python
# LOGICAL OPERATORS

ticker = ""
a = 0
if ticker: # this will be False because ticker has a faly value
    print("we have a ticker")
else:
    print("ticker is empty")

if not ticker: # this will be true because not changes falsy to Truthy
    print("we have a ticker")
else:
    print("ticker is empty") 
      
if not ticker and not a: # thsi will do same as before but both are truthy now
    print("we have a ticker")
else:
    print("ticker is empty") 

if not ticker and a: # Now one is truthy and other is falsy, so overall falsy
    print("we have a ticker")
else:
    print("ticker is empty") 

# for an and to be tru both need to be tru, if one or both false it is false
```
