```python
#Module 1.3
from decimal import Decimal
a = 0.1
b = 0.2

#0.1+0.2 != 0.3
# Python stores the decimals as bbinary
# sometimes these decimals cannot be represented by a 1 or 0
# Just like in math, 1/3= 0.333...

print(a+b)

print(a + b == 0.3)

# The Decimal function from decimal mkodulen keeps the exact float 
# in Print we used Decimal("0.3") because 0.3 would not be exactly 0.3
# same issue as before in decimals and pythion
#Decimal("0.3") gives exactly 0.3
a1 = Decimal("0.1")
b1 = Decimal("0.2")

print(a1+b1)

print(a1 + b1 == Decimal("0.3"))
```

```python
# Arithematic Modules

shares = 7
total_cost = 500
avg_price = total_cost/7
print(avg_price)

#Floor dision: first rounds down, then only keeps whole numbers
whole_lots = total_cost//shares

#Modulo, only keeps remainder. 7*71=497
#it means after buying 71 whole shares we will have 3$ left
remainder = total_cost%shares

print("Whole lots:", whole_lots, "and remainder:", remainder)

```

```python
#Daily class assignment
price1 = 102.50
price2 = 104.75
price3 = 101.20
price4 = 105.90
price5 = 103.15

highest = max(price1, price2, price3, price4, price5)
lowest = min(price1, price2, price3, price4, price5)

average = (price1+ price2+ price3+ price4+ price5)/5

print(f"Highest: {highest:.2f}")
print(f"Lowest: {lowest:>12.2f}")
print(f"Average: {average:.2f}")




```
