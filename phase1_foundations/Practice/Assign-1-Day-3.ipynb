Invalid data — a price did print, but it's nonsense: zero, or negative. A real closing price is never ≤ 0; if you see one, it's a bad tick, not a real -100% crash.

Your job is to write one script that walks through the week day-by-day, computes each day-over-day percentage change only when it's actually safe to do so, correctly labels every day that can't be computed (and why), tallies everything up, and prints a full report — without ever crashing, and without using tools you haven't learned yet (no try/except, no loops, no lists).
