```python
daily_return = -0.012

if daily_return > 0:
    print("Up day")
else: # Else takes no condition, it covers everything not included in if
    print("Not an up day")
```

```python
daily_return = 0.0003

if daily_return > 0.001: # First condition
    classification = "up"
elif daily_return < -0.001: # Second condition
    classification = "down"
else: # Every orther state
    classification = "flat"

print(classification)   # flat
```

```python
# The First-Match-Wins Rule Python tests conditions top to bottom
# in order. The instant one evaluates True
# its block runs and every remaining elif/else in that chain is skipped entirely
```

```python
move = 0.08   # an 8% move

# WRONG ORDER
if abs(move) > 0.01:
    size = "moderate"
elif abs(move) > 0.05:
    size = "big"
else:
    size = "small"

print(size)   

# lOGIC
# 0.08 > 0.01
# 0.08 > 0.05
# Both statements are true, but only >0.01 will execute
# Because it is the first statement that is True
```

```python
# Conditional Expressions

pct_change = 0.0234

# Syntax
# value_if_true  if  Condition  else  Vlaue_if_false
label = "gain" if pct_change >= 0 else "loss"
print(label)   # gain
```
